Create the Single Layer Perceptron Class

```python

class Perceptron:

    # Step 1: Initialize the Perceptron
    def __init__(self, num_inputs, learning_rate=0.01):
        # Initialize weights (extra weight for bias)
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    # Step 2: Linear Layer (Weighted Sum)
    def linear(self, inputs):
        # Z = XW + b
        z = np.dot(inputs, self.weights[1:]) + self.weights[0]
        return z

    # Step 3: Activation Function (Heaviside Step Function)
    def heaviside_step_fn(self, z):
        if z >= 0:
            return 1
        else:
            return 0

    # Step 4: Prediction
    def predict(self, inputs):
        z = self.linear(inputs)
        return self.heaviside_step_fn(z)

    # Step 5: Loss (Prediction Error)
    def loss(self, prediction, target):
        return target - prediction

    # Step 6: Train on One Sample
    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)

        # Update weights
        self.weights[1:] += self.learning_rate * error * inputs

        # Update bias
        self.weights[0] += self.learning_rate * error

    # Step 7: Fit the Model
    def fit(self, X, y, num_epochs=100):
        for epoch in range(num_epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)

    # Test the model
    def predict_batch(self, X):
        return np.array([self.predict(x) for x in X])
```

Task 1 - Implement AND Logic Gate

```python
import numpy as np

# -----------------------------
# Perceptron Class
# -----------------------------
class Perceptron:

    def __init__(self, num_inputs, learning_rate=0.1):
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    def linear(self, inputs):
        return np.dot(inputs, self.weights[1:]) + self.weights[0]

    def heaviside_step_fn(self, z):
        if z >= 0:
            return 1
        return 0

    def predict(self, inputs):
        return self.heaviside_step_fn(self.linear(inputs))

    def loss(self, prediction, target):
        return target - prediction

    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)

        # Update weights
        self.weights[1:] += self.learning_rate * error * inputs

        # Update bias
        self.weights[0] += self.learning_rate * error

    def fit(self, X, y, num_epochs=100):
        for epoch in range(num_epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)

    def predict_batch(self, X):
        return np.array([self.predict(x) for x in X])


# -----------------------------
# AND Gate
# -----------------------------
X_and = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_and = np.array([0,0,0,1])

and_model = Perceptron(num_inputs=2, learning_rate=0.1)

# Increase epochs to improve accuracy
and_model.fit(X_and, y_and, num_epochs=100)

predictions = and_model.predict_batch(X_and)

accuracy = np.mean(predictions == y_and) * 100

print("AND Gate")
print("Predictions:", predictions)
print("Accuracy:", accuracy,"%")
```

```python
X_or = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_or = np.array([0,1,1,1])

or_model = Perceptron(num_inputs=2, learning_rate=0.1)

or_model.fit(X_or, y_or, num_epochs=100)

predictions = or_model.predict_batch(X_or)

accuracy = np.mean(predictions == y_or) * 100

print("\nOR Gate")
print("Predictions:", predictions)
print("Accuracy:", accuracy,"%")
```

Task 2 - Generate a Random Dataset

```python

# Cell 5: Task 2 - Generate a Random Dataset


np.random.seed(42)

# Generate 100 random binary samples
X_random = np.random.randint(0, 2, (100, 2))

# Target Rule:
# Output = 1 if at least one input is 1
# Otherwise Output = 0

y_random = np.where(np.sum(X_random, axis=1) >= 1, 1, 0)

print("First 10 Samples")
print(X_random[:10])

print("\nFirst 10 Labels")
print(y_random[:10])
```

Train the Perceptron Using the Random Dataset

```python

# Cell 6: Train the Perceptron Using the Random Dataset


random_model = Perceptron(num_inputs=2, learning_rate=0.1)

random_model.fit(X_random, y_random, num_epochs=200)

print("Training Completed Successfully!")
```

Predict the Output and Calculate Accuracy

```python

# Cell 7: Predict the Output and Calculate Model Accuracy

predictions = random_model.predict_batch(X_random)

accuracy = np.mean(predictions == y_random) * 100

print("Random Dataset Results")
print("----------------------")
print("Predicted Values")
print(predictions)

print("\nActual Values")
print(y_random)

print("\nModel Accuracy :", accuracy,"%")
```
