Import libraries and download NLTK data

```python
import nltk
import string
import numpy as np

from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.probability import FreqDist

from gensim.models import Word2Vec

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

print("All libraries and datasets loaded successfully!")
```

**PART A: Load Brown Corpus**

load the Brown Corpus and select a suitable text category

```python
sentences = brown.sents(categories='news')

print("Number of sentences:", len(sentences))
print("\nSample sentences:")

for sentence in sentences[:5]:
    print(" ".join(sentence))
```

Convert text to lowercase

```python
lowercase_sentences = []

for sentence in sentences:
    lowercase_sentence = [word.lower() for word in sentence]
    lowercase_sentences.append(lowercase_sentence)

print("Original sentence:")
print(" ".join(sentences[0]))

print("\nLowercase sentence:")
print(" ".join(lowercase_sentences[0]))
```

Remove punctuation and stop words

```python
stop_words = set(stopwords.words('english'))

preprocessed_sentences = []

for sentence in lowercase_sentences:
    cleaned_sentence = []

    for word in sentence:
        # Remove punctuation
        word = word.strip(string.punctuation)

        # Keep only valid words
        if word and word.isalpha() and word not in stop_words:
            cleaned_sentence.append(word)

    preprocessed_sentences.append(cleaned_sentence)

print("Original:")
print(" ".join(sentences[0]))

print("\nPreprocessed:")
print(" ".join(preprocessed_sentences[0]))
```

Lemmatization

```python
lemmatizer = WordNetLemmatizer()

lemmatized_sentences = []

for sentence in preprocessed_sentences:
    lemmatized_sentence = [
        lemmatizer.lemmatize(word)
        for word in sentence
    ]

    lemmatized_sentences.append(lemmatized_sentence)

print("Before Lemmatization:")
print(preprocessed_sentences[0])

print("\nAfter Lemmatization:")
print(lemmatized_sentences[0])
```

Display original and final preprocessed text

```python
print("===== PART A: TEXT PREPROCESSING =====")

print("\nORIGINAL TEXT:")
print(" ".join(sentences[0]))

print("\nPREPROCESSED TEXT:")
print(" ".join(lemmatized_sentences[0]))
```

**PART B — Vocabulary & Word Frequency**

Create the vocabulary

```python
# PART B: VOCABULARY AND WORD FREQUENCY ANALYSIS

# Combine all preprocessed words
all_words = []

for sentence in lemmatized_sentences:
    all_words.extend(sentence)

# Create vocabulary
vocabulary = set(all_words)

print("Vocabulary Size:", len(vocabulary))
```

Calculate word frequencies

```python
# Calculate word frequencies
freq_dist = FreqDist(all_words)

print("Total number of words:", len(all_words))
print("Number of unique words:", len(vocabulary))
```

Display the 10 most frequent words

```python
print("----- TOP 10 MOST FREQUENT WORDS -----")

for word, frequency in freq_dist.most_common(10):
    print(word, ":", frequency)
```

Display frequency distribution

```python
print("\n===== FREQUENCY DISTRIBUTION =====")

for word, frequency in freq_dist.most_common(10):
    print(f"{word:<15} {frequency}")
```

**PART C — Word2Vec Embedding Generation**

Train Word2Vec

```python
# PART C: WORD2VEC EMBEDDING GENERATION

model = Word2Vec(
    sentences=lemmatized_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=4,
    seed=42
)

print("Word2Vec model trained successfully!")
print("Embedding dimension:", model.vector_size)
print("Context window:", model.window)
print("Architecture: Skip-gram")
```

Generate an embedding vector

```python
word = "state"

vector = model.wv[word]

print("Word:", word)
print("Embedding vector:")
print(vector)

print("\nVector dimensions:", vector.shape)
```

Find 5 most similar words

```python
similar_words = model.wv.most_similar("state", topn=5)

print("===== FIVE MOST SIMILAR WORDS TO 'state' =====")

for word, similarity in similar_words:
    print(f"{word} : {similarity:.4f}")
```

**PART D — Word Similarity Analysis**

Check the word pairs

```python
word_pairs = [
    ("state", "nation"),
    ("year", "last"),
    ("new", "first")
]

for word1, word2 in word_pairs:
    print(word1, "-", word2)
    print(word1 in model.wv.key_to_index, word2 in model.wv.key_to_index)
```

Calculate cosine similarity

```python
from numpy import dot
from numpy.linalg import norm

print("===== COSINE SIMILARITY ANALYSIS =====")

similarity_scores = {}

for word1, word2 in word_pairs:
    vector1 = model.wv[word1]
    vector2 = model.wv[word2]

    similarity = dot(vector1, vector2) / (norm(vector1) * norm(vector2))

    similarity_scores[(word1, word2)] = similarity

    print(f"{word1} - {word2} : {similarity:.4f}")
```

Identify the most similar pair

```python
most_similar_pair = max(
    similarity_scores,
    key=similarity_scores.get
)

print("\n===== MOST SIMILAR WORD PAIR =====")
print(
    f"{most_similar_pair[0]} - {most_similar_pair[1]} : "
    f"{similarity_scores[most_similar_pair]:.4f}"
)
```
